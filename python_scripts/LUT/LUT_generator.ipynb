```python
import itertools

### Read files containing addresses and values
# paths
ADDR_FILE_PATH = "../../matlab/generated_data/LUT_ADDR_NEW_Q_0_6_6.txt"
VALS_FILE_PATH = "../../matlab/generated_data/LUT_VALS_NEW_Q_0_13_6.txt"
DEST_FILE_PATH = "TEST_LUT_result.txt"

# actual reading
addr_file = open(ADDR_FILE_PATH, "r")
addresses = []
for line in addr_file:
    addresses.append(line[0:-1])
addr_file.close()

vals_file = open(VALS_FILE_PATH, "r")
vals = []
for line in vals_file:
    vals.append(line[0:-1])
vals_file.close()

### Write data to LUT file
# config constants
MODULE_DEFINITION = """`timescale 1ns / 1ps
//////////////////////////////////////////////////////////////////////////////////
// Engineer: Marcin Maj
// 
// Create Date: 06.08.2023 10:39:58
// Design Name: CFD
// Module Name: LUT
// Description: module implements LUT containing results for function 1/(a1+a2)
//              where both a1 and a2 are Q(0.3.3) format vals
//              result is Q(0.7.6)
//////////////////////////////////////////////////////////////////////////////////

module LUT #(
  ADDR_WIDTH = 6,
  DATA_WIDTH = 16
)(
  input  logic                   clk    ,
  input  logic                   rst_p  ,
  input  logic  [ADDR_WIDTH-1:0] address,
  output logic  [DATA_WIDTH-1:0] data_o
);
"""
MODULE_END = "endmodule"
ADDR_IN_NAME = "address"
ADDR_VAR_NAME = "address_internal"
OUT_SIG_NAME = "data_o"
PROCESS_START = "always_ff @(posedge clk) begin\n"
CASE_START = f'\tcase({ADDR_IN_NAME})\n'
CASE_END = "\tendcase\n"
PROCESS_END = "end\n"
ADDR_BITS = "12"
ADDR_FORMAT = "\'b"
RESULT_BITS = "13"
RESULT_FORMAT = "\'h"

# actual writing to file
lut_file = open(DEST_FILE_PATH, "w")
lut_file.write(MODULE_DEFINITION)
lut_file.write("\n")
lut_file.write(PROCESS_START)
lut_file.write(CASE_START)
for av_obj in zip(addresses, vals):
    lut_file.write(f'\t\t{ADDR_BITS}{ADDR_FORMAT}{av_obj[0]} : {OUT_SIG_NAME} <= {RESULT_BITS}{RESULT_FORMAT}{av_obj[1]};\n')
lut_file.write(CASE_END)
lut_file.write(PROCESS_END)
lut_file.write(MODULE_END)
lut_file.close()
```
